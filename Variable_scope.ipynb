------------------------------------------------------------
📌 What is Scope?
------------------------------------------------------------

Scope means the area where a variable can be accessed.

Simply,

Scope = Where a variable is visible and usable.

There are two main types:

1. Local Scope
2. Global Scope

------------------------------------------------------------
🟢 Local Variable
------------------------------------------------------------

Definition:

A local variable is a variable created inside a function.

Rules:

✅ Can be accessed only inside the function.
✅ Can be modified inside the function.
❌ Cannot be accessed outside the function.

Example:

def demo():
    name = "Ram"
    print(name)

demo()

Output:
Ram

------------------------------------------------------------
❌ Accessing a Local Variable Outside
------------------------------------------------------------

Example:

def demo():
    name = "Ram"

demo()

print(name)

Output:

NameError: name 'name' is not defined

Reason:

'name' is a local variable.
It exists only inside the function.

------------------------------------------------------------
🟢 Modifying a Local Variable
------------------------------------------------------------

Example:

def demo():
    age = 22
    age = 23
    print(age)

demo()

Output:

23

No keyword is required because the variable belongs to the current function.

------------------------------------------------------------
🌍 Global Variable
------------------------------------------------------------

Definition:

A global variable is created outside all functions.

Rules:

✅ Can be accessed outside the function.
✅ Can also be accessed inside the function.
❌ Cannot be modified inside a function without using the global keyword.

Example:

name = "Python"

def demo():
    print(name)

demo()

print(name)

Output:

Python
Python

------------------------------------------------------------
🌍 Modifying a Global Variable
------------------------------------------------------------

Example:

count = 10

def demo():
    global count
    count = count + 1

demo()

print(count)

Output:

11

Explanation:

The 'global' keyword tells Python:

"Use the global variable instead of creating a local one."

------------------------------------------------------------
❌ Without global Keyword
------------------------------------------------------------

Example:

count = 10

def demo():
    count = count + 1

demo()

Output:

UnboundLocalError

Reason:

Python thinks 'count' is a local variable because it is being assigned inside the function.

------------------------------------------------------------
🟢 Modifying a Global Variable Outside a Function
------------------------------------------------------------

Example:

count = 10

count = count + 1

print(count)

Output:

11

Explanation:

Since we are already in the global scope,
no 'global' keyword is required.

------------------------------------------------------------
🎯 Golden Rules
------------------------------------------------------------

✔ Local Variable

• Created inside a function.
• Accessible only inside the function.
• Can be modified inside the function.
• Cannot be accessed outside the function.

✔ Global Variable

• Created outside all functions.
• Accessible inside and outside functions.
• Can be modified directly outside functions.
• To modify it inside a function, use the global keyword.

------------------------------------------------------------
🧠 Interview Definition
------------------------------------------------------------

Scope:

Scope is the area where a variable can be accessed.

Local Variable:

A local variable is created inside a function.
It can only be accessed and modified within that function.

Global Variable:

A global variable is created outside all functions.
It can be accessed throughout the program.
To modify it inside a function, we use the global keyword.

------------------------------------------------------------
📌 Memory Trick
------------------------------------------------------------

🏠 Local Variable

Lives only inside its own function.

Outside the function → NameError

🌍 Global Variable

Lives outside all functions.

Accessible everywhere.

Modify outside → No keyword.

Modify inside a function → Use global keyword.
