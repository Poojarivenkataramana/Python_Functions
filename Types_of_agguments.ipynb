# What are Positional Arguments?

Positional arguments are matched with parameters based on their position.

Python matches the first argument with the first parameter, the second argument with the second parameter, and so on.

## Example

```python
def student(name, age):
    print(name)
    print(age)

student("Ram", 22)
```

Python internally does:

```python
name = "Ram"
age = 22
```

Output

```text
Ram
22
```

# Positional Arguments - Order Matters

In positional arguments, the order of arguments is very important.

Example

```python
def student(name, age):
    print(name)
    print(age)

student(22, "Ram")
```

Python internally does:

```python
name = 22
age = "Ram"
```

Output

```text
22
Ram
```

Python does not know that 22 is an age.

It only matches values from left to right.

# What are Keyword Arguments?

Keyword arguments match values using the parameter names instead of their position.

Because of this, the order of arguments does not matter.

## Example

```python
def student(name, age):
    print(name)
    print(age)

student(age=22, name="Ram")
```

Python internally does:

```python
name = "Ram"
age = 22
```

Output

```text
Ram
22
```

# More Examples of Keyword Arguments

Example

```python
def employee(name, salary):
    print(name)
    print(salary)

employee(salary=50000, name="Rahul")
```

Output

```text
Rahul
50000
```

Notice that the order is different, but the output is still correct because Python matches using parameter names.

# What are Default Arguments?

A default argument is a parameter that already has a predefined value.

If the user does not pass any value, Python uses the default value automatically.

## Example

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ram")
```

Output

```text
Hello Guest
Hello Ram
```

# How Default Arguments Work

Example

```python
def country(place="India"):
    print(place)

country()
country("Japan")
```

Python internally does:

First Call

```python
place = "India"
```

Second Call

```python
place = "Japan"
```

Output

```text
India
Japan
```

# Important Rule for Default Arguments

Normal parameters must always come before default parameters.

✅ Correct

```python
def student(name, age=18):
    print(name, age)
```

❌ Wrong

```python
def student(age=18, name):
    print(name, age)
```

This produces a SyntaxError because a default parameter cannot come before a normal parameter.

# Comparison Table

| Argument Type | Matches By | Order Matters |
|---------------|------------|---------------|
| Positional | Position | Yes |
| Keyword | Parameter Name | No |
| Default | Default Value | Optional |

## Quick Reminder

Positional → Match by position

Keyword → Match by parameter name

Default → Uses predefined value when no argument is passed

# Interview Definitions

## Keyword Argument

Arguments are matched using parameter names, so the order of arguments does not matter.
