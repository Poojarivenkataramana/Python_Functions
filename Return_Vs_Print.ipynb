# What is the Return Statement?

The `return` statement is used to send a value back from a function to the place where the function was called.

Unlike `print()`, which only displays the result on the screen, `return` allows us to store and reuse the result.

## Example

```python
def add():
    return 10 + 20
```

The function returns the value `30`.

# Difference Between print() and return

## Explanation of the previous logical error

The original `grade_calculator` function had a logical flaw in its `elif` statements. For instance, a condition like `elif 80 >= marks < 90` might seem to mean "if marks are between 80 and 89 (inclusive of 80)". However, in Python, this is evaluated as `(80 >= marks) and (marks < 90)`.

Consider `marks = 85`:
- `80 >= 85` is `False`.
- `85 < 90` is `True`.
- `False and True` results in `False`.

This means that for a score of 85, the condition `80 >= marks < 90` would be `False`, and the student would not receive an 'A' grade. Instead, the program would continue checking subsequent `elif` conditions, potentially assigning an incorrect grade or even 'Fail' if no other `elif` condition was met.

The corrected code checks the conditions in descending order (`marks >= 90`, `marks >= 80`, etc.), so that the first condition met is the correct one, ensuring that `marks = 85` correctly falls into the `elif marks >= 80` block to receive an 'A' grade.

````python
# 🧠 Challenge 12 – Palindrome Number

# Create

# def palindrome(num):
def palindrome(num):
    s = str(num)  # Convert to string to support indexing and len()
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return "Not Palindrome"
        left += 1
        right -= 1  # Move inward!

    return "Palindrome"

# Test cases
print(palindrome(121))    # Palindrome
print(palindrome("madam")) # Palindrome
print(palindrome(123))    # Not Palindrome





# If the number is a palindrome, return:

# Palindrome

# Otherwise return:

# Not Palindrome
````

````python
def palindrome(num):
    s = str(num)
    if s == s[::-1]:
      return 'Palindrome'
    else:
      return "Not a Palindrome"

print(palindrome(121))    # Palindrome
print(palindrome("madam")) # Palindrome
print(palindrome(123))    # Not Palindrome

````
