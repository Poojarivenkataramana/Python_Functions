# Lesson 3 - Step 1: What is `def`?

Before creating a function, Python needs to know that we are about to define one.

For this purpose, Python provides a special keyword called **`def`**.

## Definition

`def` is a **keyword** used to define (create) a function.

## Real-Life Example

Imagine you are talking to Python.

You say:

> "I am going to create a function."

How does Python know that?

It knows because you start with the keyword **`def`**.

Just like a teacher says,

> "Open your notebooks."

Students immediately understand it is an instruction.

Similarly,

`def` tells Python:

> "The next block of code is a function."

## Example

```python
def greet():
```

When Python sees `def`, it immediately understands that a new function is being created.

## Conclusion

Always use the keyword **`def`** whenever you want to create a function.

# Lesson 3 - Step 2: Function Name

Every function must have a name.

The function name helps Python identify and execute the correct function whenever it is called.

## Example

```python
def greet():
```

Here,

**greet** is the function name.

## Real-Life Example

Think about your mobile phone contacts.

Every contact has a name.

Without names, you wouldn't know whom to call.

Functions work the same way.

Without a name, Python cannot identify which function should be executed.

## Good Function Names

```python
login()

calculate_salary()

send_email()

withdraw_money()
```

These names clearly describe what the function does.

## Poor Function Names

```python
abc()

xyz()

test123()

a()
```

These names do not clearly explain the purpose of the function.

## Conclusion

Always choose meaningful function names that describe the task performed by the function.

# Lesson 3 - Step 3: Parentheses ()

Notice the function definition.

```python
def greet():
```

Even though nothing is written inside the parentheses, they are still required.

## Why do we use parentheses?

Parentheses are used to hold **inputs** that a function may receive.

These inputs are called **parameters**.

At this stage, the parentheses are empty because the function does not require any input.

Example:

```python
def greet():
```

Later, we will learn functions with parameters.

Example:

```python
def add(a, b):
```

Here,

`a` and `b` are parameters placed inside the parentheses.

## Conclusion

Every function definition must include parentheses.

Sometimes they are empty.

Sometimes they contain parameters.

# Lesson 3 - Step 4: Colon (:)

The colon (`:`) tells Python that the function body begins on the next line.

## Correct Example

```python
def greet():
```

After the colon, the function body starts.

Example:

```python
def greet():
    print("Hello")
```

## Incorrect Example

```python
def greet()
```

Python raises a **SyntaxError** because the colon is missing.

## Conclusion

Always place a colon (`:`) at the end of the function definition.

# Lesson 3 - Step 5: Indentation

After writing the function definition, the statements inside the function must be indented.

Python uses indentation to identify which statements belong to the function.

## Correct Example

```python
def greet():
    print("Hello")
    print("Welcome")
```

Both statements belong to the function because they are indented.

## Incorrect Example

```python
def greet():
print("Hello")
```

Python raises an **IndentationError** because the statement is not indented.

## Real-Life Example

Imagine a classroom.

Only the students inside the classroom belong to that class.

Similarly,

Only the statements inside the indentation belong to the function.

## Conclusion

Indentation is mandatory in Python.

Without proper indentation, Python cannot determine which statements belong to the function.

# Lesson 3 - Summary

Consider the following function:

```python
def greet():
    print("Hello")
```

Let's understand each part.

| Part | Meaning |
|------|---------|
| `def` | Defines (creates) a function |
| `greet` | Function name |

| `:` | Indicates the start of the function body |
| Indentation | Statements that belong to the function |

A function definition consists of:
