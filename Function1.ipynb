# Step 1: First Understand the Problem

Imagine your mother asks you:

**"Make tea."**

You go to the kitchen and perform these steps:

1. Boil water.
2. Add tea powder.
3. Add sugar.
4. Add milk.
5. Serve the tea.

After one hour, your father also asks:

**"Make tea."**

Will you learn everything again?

**No.**

You simply repeat the same steps.

Now imagine 10 people ask for tea.

Will you write the same 5 steps 10 different times?

**No.**

Instead, you create one process called:

**Make Tea**

Whenever someone says **"Make Tea"**, you perform those same steps.

### Conclusion

A **function** is exactly like this.

It is a **single process** that contains a set of steps.

Whenever we need those steps again, we simply call the function instead of writing everything again.

# Step 5: What is a Function? (Easy Definition)

## Definition

A **function** is a **named block of code** that performs a **specific task**.

Let's understand this definition one part at a time.

### 1. Named

Every function has a name.

Example:

```python
def greet():
    print("Hello")
```

Here, **greet** is the function's name.

---

### 2. Block of Code

A function contains one or more statements that work together.

Example:

```python
def greet():
    print("Hello")
    print("Welcome")
```

Both print statements belong to the same function.

---

### 3. Specific Task

A function should perform one particular job.

Examples:

- greet() → Greets the user.
- add() → Adds two numbers.
- login() → Logs in a user.
- calculate_salary() → Calculates salary.

Each function should focus on one task.

---

## Conclusion

A function is a reusable block of code with a name that performs one specific task whenever we call it.

# Step 7: Creating a Function vs Calling a Function

Many beginners make one common mistake.

They think that writing a function automatically executes it.

It does **not**.

## Creating a Function

```python
def greet():
    print("Hello")
```

This only creates the function.

Python remembers it, but it does not execute it.

So the output is:

```
No Output
```

---

## Calling a Function

To execute the function, we must call it.

```python
def greet():
    print("Hello")

greet()
```

Output:

```
Hello
```

---

## Real-Life Example

Imagine you buy a washing machine.

Does it start washing clothes as soon as you bring it home?

**No.**

You must press the **Start** button.

Similarly,

Creating a function is like buying the washing machine.

Calling the function is like pressing the Start button.

Only then does the work begin.

---

## Conclusion

- `def greet():` → Creates the function.
- `greet()` → Executes (calls) the function.

Remember:

**Creating is not executing.**

# Lesson 2 - Step 1: Why Do We Need Functions?

Imagine you are a software developer working for a bank.

Your manager asks you to display the following message whenever a customer logs in:

- Welcome to ABC Bank
- Have a Nice Day

Without using functions, you might write:

```python
print("Welcome to ABC Bank")
print("Have a Nice Day")

print("Welcome to ABC Bank")
print("Have a Nice Day")

print("Welcome to ABC Bank")
print("Have a Nice Day")
```

Notice something?

The same code is written again and again.

This is called **Code Duplication**.

Writing the same code repeatedly makes the program:

- Longer
- Harder to read
- Harder to maintain
- More likely to contain mistakes

## Solution

Instead of writing the same code many times, we create a function.

```python
def welcome():
    print("Welcome to ABC Bank")
    print("Have a Nice Day")
```

Now, whenever a customer logs in, we simply call:

```python
welcome()
```

One function can be used thousands of times.

## Conclusion

Functions help us **write code once and reuse it whenever needed.**
